```python
from pathlib import Path
import pandas as pd
import numpy as np

DATA_DIR = Path("../data/raw")

customers = pd.read_csv(DATA_DIR / "olist_customers_dataset.csv")
orders = pd.read_csv(DATA_DIR / "olist_orders_dataset.csv")
order_items = pd.read_csv(DATA_DIR / "olist_order_items_dataset.csv")
payments = pd.read_csv(DATA_DIR / "olist_order_payments_dataset.csv")
reviews = pd.read_csv(DATA_DIR / "olist_order_reviews_dataset.csv")
products = pd.read_csv(DATA_DIR / "olist_products_dataset.csv")
sellers = pd.read_csv(DATA_DIR / "olist_sellers_dataset.csv")
geolocation = pd.read_csv(DATA_DIR / "olist_geolocation_dataset.csv")
category_translation = pd.read_csv(DATA_DIR / "product_category_name_translation.csv")
```

```python
tables = {
    "customers": customers,
    "orders": orders,
    "order_items": order_items,
    "payments": payments,
    "reviews": reviews,
    "products": products,
    "sellers": sellers,
    "geolocation": geolocation,
    "category_translation": category_translation
}

table_summary = pd.DataFrame([
    {
        "table": name,
        "rows": df.shape[0],
        "columns": df.shape[1]
    }
    for name, df in tables.items()
])

table_summary
```

```python
missing_summary = pd.DataFrame([
    {
        "table": name,
        "column": col,
        "missing_count": df[col].isna().sum(),
        "missing_pct": round(df[col].isna().mean() * 100, 2)
    }
    for name, df in tables.items()
    for col in df.columns
    if df[col].isna().sum() > 0
])

missing_summary.sort_values(
    ["missing_pct", "missing_count"],
    ascending=False
)
```

## Missing Value Observations

- `review_comment_title` has a very high percentage of missing values (88.34%). This may indicate that many customers did not provide a review title.
- `review_comment_message` has a high percentage of missing values (58.70%). This may indicate that many customers did not provide written comments.
- `order_delivered_customer_date` has 2.98% missing values. The reason for these missing values needs to be investigated using `order_status`.
- `order_delivered_carrier_date` has 1.79% missing values. The reason for these missing values should also be investigated using order status and other order timestamps.
- Several product attributes have 1.85% missing values. These include category, name length, description length, and photo count. The reason for these missing values needs further investigation.
- Product dimension fields have very few missing values (0.01%), so their impact is likely small, but they should still be documented.

```python
duplicate_summary = pd.DataFrame([
    {
        "table": name,
        "duplicate_rows": df.duplicated().sum(),
        "duplicate_pct": round(df.duplicated().mean() * 100, 2)
    }
    for name, df in tables.items()
])

duplicate_summary.sort_values(
    "duplicate_rows",
    ascending=False
)
```

## Duplicate Value Observations

- The `geolocation` table contains **261,831 completely duplicated rows**.
- `geolocation_zip_code_prefix` is not unique, with **19,015 unique ZIP prefixes** across the table.
- A single ZIP prefix can therefore have multiple geolocation records.
- The duplicate records will **not be removed at this stage**. Their treatment will be decided during the data cleaning phase after understanding how the table will be used.

```python
key_checks = {
    "customers.customer_id": customers["customer_id"].nunique(),
    "customers.customer_unique_id": customers["customer_unique_id"].nunique(),
    "orders.order_id": orders["order_id"].nunique(),
    "order_items.order_id": order_items["order_id"].nunique(),
    "order_items.product_id": order_items["product_id"].nunique(),
    "sellers.seller_id": sellers["seller_id"].nunique(),
    "products.product_id": products["product_id"].nunique(),
}

key_checks
```
